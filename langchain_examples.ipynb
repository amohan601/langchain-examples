### RunnablePassThrough

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai.chat_models import ChatOpenAI
```

```python
chat_template_places = ChatPromptTemplate.from_template('''
Suggest only list of places in {country} to visit this summer.
''')
chat_template_favoritefood = ChatPromptTemplate.from_template('''
Suggest the best food to eat in these {places}.
''')

model =ChatOpenAI(model = 'gpt-5.4-mini', temperature = 0, max_tokens = 200, seed = 356)
output_parser = StrOutputParser()
```

```python
chain_places = chat_template_places | model | output_parser
```

```python
output_places = chain_places.invoke({'country': 'usa'})
```

```python
print(output_places)
```

```python
from langchain_core.runnables import chain
```

```python
@chain
def runnable_sum(x):
    return sum(x)

@chain
def runnable_cube(x):
    return x^3
```

```python
type(runnable_sum), type(runnable_cube)
```

```python
chain = runnable_sum | runnable_cube
```

```python
output = chain.invoke([1,2,3])
print(output)
```
